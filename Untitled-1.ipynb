```python
df_clean = df.copy()

# 1. Remove duplicates
df_clean = df_clean.drop_duplicates()

# 2. Fix text casing
text_cols = ['fuel_type', 'transmission', 'seller_type', 'insurance_valid', 'owner_type', 'body_type', 'city', 'state']
for col in text_cols:
    df_clean[col] = df_clean[col].str.strip().str.title()

# 3. Fix zero prices
df_clean['price_inr'] = df_clean['price_inr'].replace(0, np.nan)
df_clean = df_clean.dropna(subset=['price_inr'])

# 4. Fill remaining missing values simply
df_clean['mileage_kmpl'] = df_clean['mileage_kmpl'].fillna(df_clean['mileage_kmpl'].median())
df_clean['engine_cc'] = df_clean['engine_cc'].fillna(df_clean['engine_cc'].median())
df_clean['seats'] = df_clean['seats'].fillna(df_clean['seats'].mode()[0])
df_clean['color'] = df_clean['color'].fillna('Unknown')
df_clean['variant'] = df_clean['variant'].fillna('Unknown')
df_clean['insurance_valid'] = df_clean['insurance_valid'].fillna('Unknown')

print(df_clean.isna().sum())
print(df_clean.shape)
```

```python
#Distribution of Car Prices 
plt.figure(figsize=(8,5))
sns.histplot(df_clean['price_inr'], bins=30, kde=True)
plt.title("Distribution Of Car Prices")
plt.xlabel("Price (INR)")
plt.ylabel("Number of Cars")
plt.show()
```

```python
#Number of Cars by Fuel Type
plt.figure(figsize=(7,5))
sns.countplot(data=df_clean, x='fuel_type', order=df_clean['fuel_type'].value_counts().index)
plt.title("Number of Cars By Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Count")
plt.show()
```

```python
#Price by Fuel Type 
plt.figure(figsize=(8,5))
sns.boxplot(data=df_clean, x='fuel_type', y='price_inr')
plt.title("Price Distribution by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Price (INR)")
plt.show()
```

```python
#kms Driven vs Price
plt.figure(figsize=(8,5))
sns.scatterplot(data=df_clean, x='kms_driven' , y='price_inr',alpha=0.5)
plt.title("Kms Driven vs Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Price (INR)")
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(6,5))
sns.heatmap(df_clean[['price_inr','kms_driven','mileage_kmpl','engine_cc']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
#Does the car's age affect its price?
plt.figure(figsize=(8,5))
sns.scatterplot(data=df_clean, x='manufacture_year', y='price_inr', alpha=0.5)
plt.title("Car Price vs Manufacture Year")
plt.xlabel("Manufacture Year")
plt.ylabel("Price (INR)")
plt.show()
```

```python
#Which brand holds its values best?
plt.figure(figsize=(10,6))
avg_price_by_brand = df_clean.groupby('brand')['price_inr'].mean().sort_values(ascending=False)
sns.barplot(x=avg_price_by_brand.values, y=avg_price_by_brand.index)
plt.title("Average Price by Brand")
plt.xlabel("Average Price (INR)")
plt.ylabel("Brand")
plt.show()
```

```python
#Does the number of previous owners affect price
plt.figure(figsize=(8,5))
avg_price_by_owner = df_clean.groupby('owner_type')['price_inr'].mean().sort_values(ascending=False)
sns.barplot(x=avg_price_by_owner.index, y=avg_price_by_owner.values)
plt.title("Average Price by Owner History")
plt.xlabel("Owner Type")
plt.ylabel("Average Price (INR)")
plt.xticks(rotation=20)
plt.show()
```

```python
#Do automatic car cost more than manual ones?
plt.figure(figsize=(7,5))
sns.boxplot(data=df_clean, x='transmission', y='price_inr')
plt.title("Price By Transmission Type")
plt.xlabel("Transmission")
plt.ylabel("Price (INR)")
plt.show()
```

```python
#Which cities have the most active used Car Market
plt.figure(figsize=(10,6))
city_counts = df_clean['city'].value_counts()
sns.barplot(x=city_counts.values, y=city_counts.index)
plt.title("Number of Listings by City")
plt.xlabel("Number of Listings")
plt.ylabel("City")
plt.show()
```

```python
#what share of the market does each fuel type hold?
plt.figure(figsize=(7,7))
fuel_counts = df_clean['fuel_type'].value_counts()
plt.pie(fuel_counts, labels=fuel_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title("Market Share by Fuel Type")
plt.show()
```

```python
#How does price distribution differ among top brands?
top_brands = ['Maruti Suzuki', 'Hyundai', 'Tata', 'Toyota']
df_top_brands = df_clean[df_clean['brand'].isin(top_brands)]

plt.figure(figsize=(10,6))
sns.histplot(data=df_top_brands, x='price_inr', hue='brand', bins=30, kde=True, alpha=0.4)
plt.title("Price Distribution: Selected Brands")
plt.xlabel("Price (INR)")
plt.ylabel("Number of Cars")
plt.show()
```
